```python
# B1
my_list = [2.72, 3.14, 12, 55, 9, 100, 12, 11, 120, 80, 1.1, 4.2, 12]
for x in [x * 2 for x in my_list]:
    print(x)
```

```python
# B2
computed_sum = 0
for x in my_list:
    computed_sum += x

print(computed_sum)
```

```python
# B3
computed_sum = 0
for x in [x for x in my_list if x > 50]:
    computed_sum += x

print(computed_sum)
```

```python
# B4
computed_product = 1
for x in [x for x in my_list if x != 12]:
    computed_product *= x

print(computed_product)
```

```python
# B5
def positive_list_sum(list1) -> int:
    positive_sum = 0
    for y in [y for y in list1 if y > 0]:
        positive_sum += y
    
    return positive_sum

a = [-1, 4, -43, 3.14, -266, 3, 900]
b = list(range(-5, 5))

print(positive_list_sum(a))
print(positive_list_sum(b))
```
