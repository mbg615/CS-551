```python
# Q2
import numpy as np

def sepia(input_img):
    sepia_filter = np.array([
        [0.393, 0.769, 0.189],
        [0.349, 0.686, 0.168],
        [0.272, 0.534, 0.131]
    ])
    sepia_img = input_img.dot(sepia_filter.T)
    sepia_img /= sepia_img.max()
    return sepia_img

demo = gr.Interface(
    fn=sepia, 
    inputs=gr.Image(), 
    outputs="image"
)

demo.launch()
```

```python
# Q4
import pandas as pd
import random

df = pd.DataFrame({
    'height': np.random.randint(120, 200, 25),
    'weight': np.random.randint(45, 150, 25),
    'age': np.random.randint(18, 65, 25),
    'major': [random.choice(["computer science", "political science", "nursing", "mathematics"]) for _ in range(25)]
})

with gr.Blocks() as demo:
    gr.LinePlot(df, x="weight", y="height")
    gr.ScatterPlot(df, x="weight", y="height")
    gr.BarPlot(df, x="weight", y="height")
    
demo.launch()
```
